# Posts

Load `posts.csv`, cast `published_at_date` to a datetime, check for nulls and duplicates, then export the cleaned data.

```python
df_posts = pd.read_csv(os.path.join(raw_data_dir, 'posts.csv'))
df_posts.head()
```

```python
df_posts.info()
```

```python
# convert published_at_date to datetime (source format is ISO YYYY-MM-DD)
df_posts['published_at_date'] = pd.to_datetime(df_posts['published_at_date'], format='%Y-%m-%d')

df_posts.info()
```

```python
df_posts.shape
```

```python
df_posts.isnull().sum()
```

```python
df_posts.duplicated().sum()
```

```python
df_posts.head()
```

```python
df_posts_cleaned = df_posts.copy()
df_posts_cleaned.to_csv(os.path.join(processed_data_dir, 'posts_cleaned.csv'), index=False)
```

# Post Stats

Load `poststats.csv`, cast `data_date` to a datetime and `views`/`watchtime` to integers (stripping thousands separators), check for nulls and duplicates, then export the cleaned data.

```python
df_post_stats = pd.read_csv(os.path.join(raw_data_dir, 'poststats.csv'))
df_post_stats.head()
```

```python
df_post_stats.info()
```

```python
# convert data_date to datetime DD/MM/YYYY
df_post_stats['data_date'] = pd.to_datetime(df_post_stats['data_date'], format='%d/%m/%Y')

# convert views to integer and strip commas
df_post_stats['views'] = df_post_stats['views'].astype(str).str.replace(',', '', regex=False).astype(int)

# convert watchtime to integer and strip commas
df_post_stats['watchtime'] = df_post_stats['watchtime'].astype(str).str.replace(',', '', regex=False).astype(int)

df_post_stats.info()
```

```python
df_post_stats.isnull().sum()
```

```python
df_post_stats.duplicated().sum()
```

```python
df_post_stats.head()
```

```python
df_post_stats_cleaned = df_post_stats.copy()
df_post_stats_cleaned.to_csv(os.path.join(processed_data_dir, 'poststats_cleaned.csv'), index=False)
```
